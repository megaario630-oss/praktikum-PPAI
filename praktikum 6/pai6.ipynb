```python

# Program menghitung bilangan genap dan ganjil

angka = (1, 2, 3, 4, 5, 6, 7, 8, 9)

genap = 0
ganjil = 0

for nilai in angka:
    if nilai % 2 == 0:
        genap += 1
    else:
        ganjil += 1

print("list angka, ", angka)
print("Jumlah bilangan genap :", genap)
print("Jumlah bilangan ganjil :", ganjil)
```
