```python

# Program menentukan jenis segitiga

x = 10
y = 10
z = 10

if x == y and y == z:
    print(f"Segitiga sama sisi, dimana memiliki nilai sisi x={x}, y={y} z={z}")

elif x == y or x == z or y == z:
    print(f"Segitiga sama kaki, dimana memiliki nilai sisi x={x}, y={y} z={z}")

else:
    print(f"Segitiga Sembarang, dimana memiliki nilai sisi x={x}, y={y} z={z}")
```
