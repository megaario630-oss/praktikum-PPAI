```python

kalimat = "PRAKTIKUM ALPRO1"

print('String seluruhnya:', kalimat)
print('Menampilkan hanya 1 huruf pertama dari string, yaitu', kalimat[0])
print('Menampilkan huruf terakhir dari string yaitu', kalimat[-1])
print('Menampilkan huruf ke-2, ke-3, ke-5, dan ke-14, antara lain:')
print('Huruf ke-2:', kalimat[1])
print('Huruf ke-3:', kalimat[2])
print('Huruf ke-5:', kalimat[4])
print('Huruf ke-14:', kalimat[13])
```

```python

x = 20
y = 10

print("Nilai awal x=", x, ", y=", y)

temp = x
x = y
y = temp

print("Nilai x setelah ditukar:", x)
print("Nilai y setelah ditukar:", y)
```

```python

a = 20
b = 9

print('Nilai dari operasi bitwise "AND" antara a:20, b:9 -> (a & b) =', a & b)
print('Nilai dari operasi bitwise "OR" antara a:20, b:9 -> (a | b) =', a | b)
print('Nilai dari operasi bitwise "XOR" antara a:20, b:9 -> (a ^ b) =', a ^ b)
```

```python

print("Program Kalkulasi midpoint")

x1 = 2
y1 = 2
x2 = 4
y2 = 4

midpoint_x = (x1 + x2) / 2
midpoint_y = (y1 + y2) / 2

print("\nNilai midpoint dari garis :")
print("Nilai midpoint x :", midpoint_x)
print("Nilai midpoint y :", midpoint_y)
```

```python

angka = 1234

ribuan = angka // 1000
ratusan = (angka // 100) % 10
puluhan = (angka // 10) % 10
satuan = angka % 10

jumlah = ribuan + ratusan + puluhan + satuan

print("angka =", angka)
print("Jumlah dari angka =", jumlah)
```
